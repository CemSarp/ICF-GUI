# GPU notebook smoke test
Run as a **Notebook, batch execution** workflow. Needs `nbconvert` and `ipykernel` in the selected environment.
The executed copy is saved with the run's outputs.

```python
import json, os, platform, sys
import torch
info = {
    'host': platform.node(),
    'python': sys.version.split()[0],
    'torch': torch.__version__,
    'cuda_build': torch.version.cuda,
    'cuda_available': torch.cuda.is_available(),
    'device': torch.cuda.get_device_name(0) if torch.cuda.is_available() else None,
    'output_dir': os.environ.get('ICFCC_OUTPUT_DIR'),
}
info
```

```python
assert info['cuda_available'], 'No GPU visible: check the partition, toolchain and PyTorch build'
x = torch.ones((256, 256), device='cuda')
info['matmul_ok'] = abs(float((x @ x).sum().item()) - 256.0 ** 3) < 1.0
with open(os.path.join(os.environ.get('ICFCC_OUTPUT_DIR', '.'), 'notebook_smoke.json'), 'w') as f:
    json.dump(info, f, indent=2)
info['matmul_ok']
```
